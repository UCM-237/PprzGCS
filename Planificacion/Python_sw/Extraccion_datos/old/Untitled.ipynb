```python
import numpy as np
import pandas as pd
import os
def extraccion_datos(var, pos, ruta_datos):
    dato = []
    tiempo = []
    with open(ruta_datos, "r", encoding="utf-8") as file:
        for line in file:
            if var in line:
                campos = line.strip().split()
                if campos[2] == var:
                    dato.append(float(campos[pos]))
                    tiempo.append(float(campos[0]))
    return np.array(dato), np.array(tiempo)
```

```python
def llevar_a_t_comun(t_comun, t_var, var, check=False):
    var_interp = []
    t_var_interp = []

    i = 0  # índice en t_var
    last_value = var[0]
    last_time = t_var[0]

    for t in t_comun:
        while i < len(t_var) and t_var[i] <= t:
            last_value = var[i]
            last_time = t_var[i]
            i += 1
        var_interp.append(last_value)
        t_var_interp.append(last_time)

    return np.array(var_interp), np.array(t_var_interp)
```

```python
nombre_archivo = "25_07_16__09_18_40.data"  # nombre que recibes desde Qt
home_dir = os.path.expanduser("~")
ruta_datos = os.path.join(home_dir, "paparazzi", "var", "logs", nombre_archivo)
```

```python
profundidades, t_prof = extraccion_datos("SERIAL_COM", 15, ruta_datos)
mode, t_mode = extraccion_datos("SERIAL_COM", 3, ruta_datos)
static_control, t_static_control = extraccion_datos("STATIC_CONTROL", 3, ruta_datos)
```

```python
static_control, t_static_control = llevar_a_t_comun(t_prof, t_static_control, static_control)
```

```python
nombre_salida = "datos_profundidad.csv"
archivo_salida = os.path.join(home_dir, "Desktop", nombre_salida)
with open(archivo_salida, 'w') as f:
    # Escribir los encabezados
    f.write("tiempo,Profundidad,static,mode\n")
    # Escribir los puntos
    for i in range(len(profundidades)):
        fila = [
            t_prof[i], profundidades[i], static_control[i], mode[i]
        ]
        f.write(",".join(str(valor) for valor in fila) + "\n")
```
